```sql
SQL JOINTS

A JOIN combines rows from two or more tables using a common column Types of Joins

INNER JOIN---Returns only matching records from both tables.
LEFT JOIN---All records from LEFT table + Matching records from RIGHT table
RIGHT JOIN--All records from RIGHT table + Matching records from LEFT table
FULL OUTER JOIN---All records from Table A + All records from Table B
SELF JOIN---Joining a table with itself.
CROSS JOIN---Every row from A * Every row from B
LEFT ANTI JOIN---Rows In A Not Present In B
LEFT SEMI JOIN---Rows In A That Exist In B (Returns rows from the left table that have a match in the right table.
```
