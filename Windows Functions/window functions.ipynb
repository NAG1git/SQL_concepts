WINDOW FUNCTIONS

ROW_NUMBER()
→ Unique Sequence

RANK()
→ Same Rank, Gap Exists

DENSE_RANK()
→ Same Rank, No Gaps

PARTITION BY
→ Grouping Without Collapsing Rows

SUM() OVER()
→ Running Total

AVG() OVER()
→ Average Per Group

COUNT() OVER()
→ Count Per Group

LAG()
→ Previous Row

LEAD()
→ Next Row

FIRST_VALUE()
→ First Row Value

LAST_VALUE()
→ Last Row Value

NTILE()
→ Divide Rows Into Buckets
